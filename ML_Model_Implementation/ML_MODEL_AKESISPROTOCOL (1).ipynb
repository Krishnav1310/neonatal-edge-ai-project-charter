```python
from collections import defaultdict

# Stores previous readings of each baby
baby_history = defaultdict(list)

# Stores baseline values of each baby
baby_baseline = {}
```

```python
def update_history(baby_id, reading):
    """
    reading = {
        'heart_rate':145,
        'spo2':97,
        'resp_rate':42,
        'temperature':36.8
    }
    """

    baby_history[baby_id].append(reading)

    # Keep only last 30 readings
    if len(baby_history[baby_id]) > 30:
        baby_history[baby_id].pop(0)
```

```python
import numpy as np

def calculate_baseline(baby_id):

    history = baby_history[baby_id]

    if len(history) < 10:
        return None

    baseline = {
        "heart_rate": np.mean([x["heart_rate"] for x in history]),
        "spo2": np.mean([x["spo2"] for x in history]),
        "resp_rate": np.mean([x["resp_rate"] for x in history]),
        "temperature": np.mean([x["temperature"] for x in history])
    }

    baby_baseline[baby_id] = baseline

    return baseline
```

```python
def calculate_risk_score(current, baseline):

    score = 0

    if baseline is None:
        return 0

    if abs(current["heart_rate"] - baseline["heart_rate"]) > 15:
        score += 30

    if current["spo2"] < 95:
        score += 35

    if abs(current["resp_rate"] - baseline["resp_rate"]) > 10:
        score += 20

    if abs(current["temperature"] - baseline["temperature"]) > 1:
        score += 15

    return min(score,100)
```

```python
def generate_reason(current, baseline):

    reasons=[]

    if baseline is None:
        return reasons

    if current["heart_rate"] > baseline["heart_rate"] + 15:
        reasons.append(f"Heart Rate: {current['heart_rate']} bpm ⬆️")

    elif current["heart_rate"] < baseline["heart_rate"] - 15:
        reasons.append(f"Heart Rate: {current['heart_rate']} bpm ⬇️")


    if current["spo2"] < 95:
        reasons.append(f"SpO₂: {current['spo2']}% ⬇️")


    if current["resp_rate"] > baseline["resp_rate"] + 10:
        reasons.append(f"Respiratory Rate: {current['resp_rate']}/min ⬆️")

    elif current["resp_rate"] < baseline["resp_rate"] - 10:
        reasons.append(f"Respiratory Rate: {current['resp_rate']}/min ⬇️")


    if current["temperature"] > baseline["temperature"] + 1:
        reasons.append(f"Temperature: {current['temperature']}°C ⬆️")

    elif current["temperature"] < baseline["temperature"] - 1:
        reasons.append(f"Temperature: {current['temperature']}°C ⬇️")

    return reasons
```

```python
def determine_status(score):

    if score < 30:
        return "NORMAL"

    elif score < 70:
        return "MODERATE"

    else:
        return "CRITICAL"
```

```python
def process_baby(baby_id, current_reading):

    # Store latest reading
    update_history(baby_id, current_reading)

    # Calculate baseline
    baseline = calculate_baseline(baby_id)

    # Until enough history is available
    if baseline is None:

        return {
            "baby_id": baby_id,
            "status": "LEARNING BASELINE"
        }

    # Risk score
    risk_score = calculate_risk_score(current_reading, baseline)

    # Status
    status = determine_status(risk_score)

    output = {
        "baby_id": baby_id,
        "status": status
    }

    if status != "NORMAL":

        output["risk_score"] = risk_score

        output["reason"] = generate_reason(
            current_reading,
            baseline
        )

    return output
```

```python
# Static per-baby clinical data (set once per baby, e.g. at admission)
baby_profiles = {}

def register_baby_profile(baby_id, gender, weight_kg, apgar_score,
                           immunizations_done, reflexes_normal):
    baby_profiles[baby_id] = {
        "gender": gender,
        "weight_kg": weight_kg,
        "apgar_score": apgar_score,
        "immunizations_done": immunizations_done,
        "reflexes_normal": reflexes_normal
    }
```

```python
def calculate_trend(baby_id, window=5):
    history = baby_history[baby_id]

    if len(history) < window:
        return {"heart_rate": 0, "spo2": 0, "resp_rate": 0, "temperature": 0}

    recent = history[-window:]
    older = history[-2*window:-window] if len(history) >= 2*window else history[:window]

    def avg(readings, key):
        return np.mean([r[key] for r in readings])

    return {
        "heart_rate": avg(recent, "heart_rate") - avg(older, "heart_rate"),
        "spo2": avg(recent, "spo2") - avg(older, "spo2"),
        "resp_rate": avg(recent, "resp_rate") - avg(older, "resp_rate"),
        "temperature": avg(recent, "temperature") - avg(older, "temperature")
    }
```

```python
rf_feature_order = list(model.feature_names_in_)

def scale_for_model(row_df):
    scaled = row_df.copy()
    for col in row_df.columns:
        if col in scaler.feature_names_in_:
            idx = list(scaler.feature_names_in_).index(col)
            scaled[col] = (row_df[col] - scaler.mean_[idx]) / scaler.scale_[idx]
    return scaled

# risk_level is binary (0/1) in this dataset, not 3-class.
# Figure out which encoded value is the "higher risk" group from the data itself.
risk_group_stats = df.groupby("risk_level")[["heart_rate_bpm", "oxygen_saturation"]].mean()
high_risk_label = risk_group_stats["oxygen_saturation"].idxmin()  # lower avg SpO2 = higher risk group
print("High-risk encoded label:", high_risk_label)
print(risk_group_stats)

def rf_predict_risk(current, profile):
    row = pd.DataFrame([{
        "gender": profile["gender"],
        "weight_kg": profile["weight_kg"],
        "temperature_c": current["temperature"],
        "heart_rate_bpm": current["heart_rate"],
        "respiratory_rate_bpm": current["resp_rate"],
        "oxygen_saturation": current["spo2"],
        "apgar_score": profile["apgar_score"],
        "immunizations_done": profile["immunizations_done"],
        "reflexes_normal": profile["reflexes_normal"]
    }])[rf_feature_order]

    row_scaled = scale_for_model(row)
    proba = model.predict_proba(row_scaled)[0]

    high_risk_idx = list(model.classes_).index(high_risk_label)
    rf_score = proba[high_risk_idx] * 100
    return rf_score
```

```python
def clinical_safety_check(current):
    alerts = []

    if current["heart_rate"] > 200 or current["heart_rate"] < 80:
        alerts.append(f"Heart Rate: {current['heart_rate']} bpm — outside safe absolute range")

    if current["spo2"] < 85:
        alerts.append(f"SpO₂: {current['spo2']}% — critically low")

    if current["resp_rate"] > 80 or current["resp_rate"] < 20:
        alerts.append(f"Respiratory Rate: {current['resp_rate']}/min — outside safe absolute range")

    if current["temperature"] > 38.5 or current["temperature"] < 35.0:
        alerts.append(f"Temperature: {current['temperature']}°C — outside safe absolute range")

    return alerts
```

```python
def process_baby_v2(baby_id, current_reading, profile):
    update_history(baby_id, current_reading)
    baseline = calculate_baseline(baby_id)

    if baseline is None:
        return {"baby_id": baby_id, "status": "LEARNING BASELINE"}

    baseline_score = calculate_risk_score(current_reading, baseline)
    rf_score = rf_predict_risk(current_reading, profile)
    trend = calculate_trend(baby_id)
    clinical_alerts = clinical_safety_check(current_reading)

    final_score = round(0.5 * baseline_score + 0.5 * rf_score, 1)

    if clinical_alerts:
        final_score = 100

    status = determine_status(final_score)

    output = {"baby_id": baby_id, "status": status, "risk_score": final_score}

    if status != "NORMAL":
        reasons = generate_reason(current_reading, baseline)
        reasons.extend(clinical_alerts)

        # fallback: RF flagged risk even though no single vital crossed a hard threshold
        if not reasons:
            reasons.append(f"AI model detected an unusual combination of vitals (model confidence: {round(rf_score)}%)")

        output["reason"] = reasons

    return output
```
